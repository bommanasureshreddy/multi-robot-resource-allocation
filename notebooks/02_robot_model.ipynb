## Composition check 

```python
assert len(fleet.robots) == 12
assert len(fleet.by_type("UGV")) == 4
assert len(fleet.by_type("USV")) == 4
assert len(fleet.by_type("UAV")) == 4
print("OK: 4 UGV / 4 USV / 4 UAV confirmed.")
```

## Capability matrix

```python
import pandas as pd
rows = []
for r in fleet.robots.values():
    row = {"robot_id": r.robot_id, "type": r.robot_type, **r.capabilities}
    rows.append(row)
pd.DataFrame(rows).set_index("robot_id")
```
